```python
def convert_price(price_str):
    price_str = price_str.strip().lower()
    price = 0
    if "tỉ" in price_str:
        parts = price_str.split("tỉ")
        ty = int(parts[0].strip())
        price += ty * 1_000_000_000
        if "triệu" in parts[1]:
            trieu = int(parts[1].replace("triệu", "").strip())
            price += trieu * 1_000_000
    elif "triệu" in price_str:
        trieu = int(price_str.replace("triệu", "").strip())
        price += trieu * 1_000_000
    return price

df1["price_num"] = df1["Price"].apply(convert_price)
```

### clean df2

```python
import re

def clean_name(text):
    if not isinstance(text, str):
        return None
    text = re.sub(r"Xe\s+", "", text)  
    text = re.sub(r"\s+", " ", text) 
    return text.strip()

df2["Name"] = df2["Name"].apply(clean_name)
```

```python
import re

def split_name_price(full_name):
    if not isinstance(full_name, str):
        return None, None
    
    if " - " in full_name:
        parts = full_name.rsplit(" - ", 1)
        car = parts[0].strip()
        price_raw = parts[1].strip().lower()
        try:
            price_raw = price_raw.replace(",", ".")  
            matches = re.findall(r"(\d+(?:[\.,]?\d*)?)\s*(tỷ|triệu)", price_raw)
            price = 0
            for value, unit in matches:
                value = float(value.replace(",", ".").replace(".", "")) if "," not in value else float(value.replace(",", "."))
                if unit == "tỷ":
                    price += value * 1_000_000_000
                elif unit == "triệu":
                    price += value * 1_000_000
        except:
            price = None
        return car, price
    return full_name, None
```

```python
df2[["Name", "Price"]] = df2["Name"].apply(
    lambda x: pd.Series(split_name_price(x))
)
```
