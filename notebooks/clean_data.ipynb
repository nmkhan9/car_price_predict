```python
def convert_price(price_str):
    price_str = price_str.strip().lower()
    price = 0
    if "tỉ" in price_str:
        parts = price_str.split("tỉ")
        ty = int(parts[0].strip())
        price += ty * 1_000_000_000
        if "triệu" in parts[1]:
            trieu = int(parts[1].replace("triệu", "").strip())
            price += trieu * 1_000_000
    elif "triệu" in price_str:
        trieu = int(price_str.replace("triệu", "").strip())
        price += trieu * 1_000_000
    return price
```

```python
df1["price"] = df1["price"].apply(convert_price)
```

```python
df1.head()
```

```python
def remove_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    filtered_df = df[(df[column] >= Q1 - 1.5 * IQR) & (df[column] <= Q3 + 1.5 * IQR)]
    return filtered_df
```

```python
df1 = remove_outliers_iqr(df1, "price")
```

1. Name

```python
import re

def clean_name(text):
    if not isinstance(text, str):
        return None
    text = re.sub(r"Xe\s+", "", text)  
    text = re.sub(r"\s+", " ", text) 
    return text.strip()
```

```python
df2["name"] = df2["name"].apply(clean_name)
```

```python
def split_name_price(full_name):
    if not isinstance(full_name, str):
        return None, None
    
    if " - " in full_name:
        parts = full_name.rsplit(" - ", 1)
        car = parts[0].strip()
        price_raw = parts[1].strip().lower()
        return car, price_raw
    return full_name, None
```

```python
df2[["name", "price"]] = df2["name"].apply(
    lambda x: pd.Series(split_name_price(x))
)
```

```python
df2.head()
```

```python
df2["brand"] = df2["name"].str.strip().str.split().str[0]
```

2. Price

```python
def cv_price(price_str):
    if not isinstance(price_str, str):
        return None
    price_str = price_str.strip().lower()
    price = 0

    try:
        ty_match = re.search(r"(\d+)[\.,]?\d*\s*t[ỷỉ]", price_str)
        if ty_match:
            ty = int(ty_match.group(1))
            price += ty * 1_000_000_000

        trieu_match = re.search(r"(\d+)[\.,]?\d*\s*triệu", price_str)
        if trieu_match:
            trieu = int(trieu_match.group(1))
            price += trieu * 1_000_000

        return price if price > 0 else None

    except Exception as e:
        return None
```

```python
df2["price"] = df2["price"].apply(cv_price)
```
